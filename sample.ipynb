```python
import tkinter as tk

class LibraryAppGUI():
    def __init__(self):
        self.root = tk.Tk()

        self.root.title("Library App")

        self.label = tk.Label(self.root, text="Library App")
        self.label.grid(row=0, column=0, padx=10, pady=10)


        self.entry = tk.Entry(self.root, width=40)
        self.entry.grid(row=0, column=1, padx=10, pady=10)

        self.button = tk.Button(self.root, text="SEARCH BOOK", font=('Arial', 8,"bold"))
        self.button.grid(row=0, column=2, padx=2, pady=4)

        self.button = tk.Button(self.root, text="SEARCH AUTHOR", font=('Arial', 8, "bold"))
        self.button.grid(row=0, column=3, padx=2, pady=4)

        self.root.mainloop()

LibraryAppGUI()



```
